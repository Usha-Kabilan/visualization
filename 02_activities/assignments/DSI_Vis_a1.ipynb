```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100,50)
y2 = np.random.randint(0, 100,50)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])

fig, ax = plt.subplots(figsize=(5, 3))

fig,(ax1, ax2) = plt.subplots(ncols=2,
 nrows=1,
 figsize=(7, 3))
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2,
 nrows=1,
 figsize=(7, 3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
fig.show()
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
 ['ax2', 'ax3']],
 figsize=(7, 4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
plt.show()
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
 ['ax2', 'ax3']],
 figsize=(7, 4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label',fontsize=18)
someaxes["ax2"].set_xlabel('Another Label',fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18 )
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
 ['ax2', 'ax3']],
 figsize=(7, 4),
 layout = "constrained")
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label', fontsize=18)
someaxes["ax2"].set_xlabel('Another Label', fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=18)
```

```python
# first make our sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])
# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(7, 3))
# now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,
color = "indigo")
ax.plot(x, y2,
color = "red")
```

```python
y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
```

```python
ax.errorbar(x, #our x values
 y2, #our y values
 yerr = y2_sd,
 fmt = 'o')
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2,
 color = "red")
ax.errorbar(x,
 y2,
 yerr = y2_sd,
 fmt = "none",
 ecolor= "indigo",
 elinewidth= 4,
 capsize = 6,
 capthick= 4 )
```

```python
import plotly.graph_objects as go # 'go' is 'graph objects’
x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
```

```python
graph = go.Figure()
graph.add_trace(go.Bar(x=x1, y=y1))
```

```python
graph.update_layout(
 title="Pirate Scores",
 xaxis_title="Pirates",
 yaxis_title="Scores"
)
graph.show()
```

```python
graph = go.Figure()
graph.add_trace(go.Scatter(x=x1, y=y1, mode='markers', # we want points for a scatter plot
 marker=dict(
 size=15, # point size
 color='hotpink', # point colour
 opacity=1, # point transparency/alpha
 line=dict(width=5, color='purple') # point outline
 )))
graph.update_layout(
 title='Interactive Pirate Plot',
 xaxis_title='Pirates',
 yaxis_title='Scores',
 width=500, height=500)
```

```python
from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",
on_bad_lines='skip')
df
```

```python
# join all our text from each row from our quote column into a string
text = " ".join(each for each in df.quote)
# generate our wordcloud image
wordcloud = WordCloud(background_color="white",
colormap = 'inferno').generate(text)
# use matplotlib syntax to put our image in a figure
fig, ax = plt.subplots(figsize=(7, 3))
ax.imshow(wordcloud, # remember 'imshow' from when we added pictures to our matplotlib axes
 interpolation='bilinear') # this line helps smooth our image
ax.axis("off")
```

```python
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted
```

```python
A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])
```

```python
diagram = venn2_unweighted([A, B],
 set_labels = ('Fruits', 'Pies'),
 set_colors=("blue", "red"),
 alpha=0.5)
plt.show()
```

```python
diagram.get_label_by_id("10")\
.set_text("n".join(A - B))
diagram.get_label_by_id("11")\
.set_text("n".join(A & B))
diagram.get_label_by_id("01")\
.set_text("n".join(B - A))
```

```python
diagram = venn2_unweighted([A, B],
 set_labels = ('Fruits', 'Pies'),
 set_colors=("blue", "red"),
 alpha=0.5)
diagram.get_label_by_id("10")\
.set_text("\n".join(A - B))
diagram.get_label_by_id("11")\
.set_text("\n".join(A & B))
diagram.get_label_by_id("01")\
.set_text("\n".join(B - A))
plt.show()
```
